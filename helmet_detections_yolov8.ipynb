## Mengimpor Modul yang Diperlukan

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
from pathlib import Path
import random
import cv2
```

## Memuat dan Melatih Model YOLOv8

```python
# Memuat model YOLOv8 pretrained
model = YOLO("yolov8n.pt")  # Ganti dengan model yang diinginkan (yolov8n.pt, yolov8s.pt, dst)

# Melatih model dengan augmentasi yang sudah diaktifkan
model.train(
    data='/kaggle/input/yolo-helmet-dataset-2/data.yaml',  # Path ke file data.yaml
    epochs=250,  # Jumlah epoch pelatihan
    imgsz=640  # Ukuran gambar
)
```

## Evaluasi Model

```python
results = model.val()  # Evaluasi model menggunakan data validasi
print(results)
```

## Menyimpan Model yang Sudah Dilatih

```python
import os

# Membuat folder jika belum ada
output_dir = 'modelyolov8'
os.makedirs(output_dir, exist_ok=True)  # Membuat folder 'modelyolo' jika belum ada

model.save(f'{output_dir}/best_model.pt')  # Simpan model ke dalam folder 'modelyolo'
```

## Menggunakan Model yang Sudah Dilatih untuk Deteksi Objek

```python
# Menggunakan model yang sudah dilatih untuk deteksi objek
img = cv2.imread("/kaggle/input/yolo-helmet-dataset-2/valid/images/IMG-20250526-WA0060_jpg.rf.c954c44f6dfa97daadf969c490077b72.jpg")  # Ganti dengan path gambar Anda

model = YOLO("/kaggle/working/modelyolov8/best_model.pt")

# Melakukan deteksi objek pada gambar
results = model(img)

# Menampilkan gambar dengan bounding box (akses objek pertama dalam list)
results[0].show()  # Menampilkan gambar dengan bounding box

# Menyimpan gambar hasil deteksi
results[0].save()  # Menyimpan gambar hasil deteksi
```
